```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import joblib

import tensorflow as tf
from tensorflow.saved_model import load

from predictions import single_input_prediction, multi_input_prediction, cal_metrics
```

# Prepaid Travel Cost

```python
##prefix data
train_prefix_trace_encoded = np.load("ptc_train_prefix.npy")
val_prefix_trace_encoded = np.load("ptc_val_prefix.npy")
test_prefix_trace_encoded = np.load("ptc_test_prefix.npy")

##probability data
train_dfg_probability = pd.read_csv("ptc_train_dfg_probability.csv")
val_dfg_probability = pd.read_csv("ptc_val_dfg_probability.csv")
test_dfg_probability = pd.read_csv("ptc_test_dfg_probability.csv")

#target
train_target_encoded = np.load("ptc_train_target.npy")
val_target_encoded = np.load("ptc_val_target.npy")
test_target_encoded = np.load("ptc_test_target.npy")

#original data
#train_target = pd.read_csv("ptc_train_target_org.csv")
test_target = pd.read_csv("ptc_test_target_org.csv")
#encoder
label_encoder = joblib.load('ptc_label_encoder.joblib')
```

## Prefix Model

```python
prefix_model = tf.keras.models.load_model("ptc_prefix_model.keras")
```

```python
#Get labels
predicted_labels = single_input_prediction(prefix_model, test_prefix_trace_encoded, label_encoder)
```

```python
#get metrics
actual_label = test_target["activity"]
#micro_average, recall, macro_recall, precision, macro_precision = cal_metrics(actual_label,predicted_labels)
macro_accuracy, recall, macro_recall, precision, macro_precision, micro_accuracy = cal_metrics(actual_label,predicted_labels)
```

```python
result = {"prefix":{"macro_accuracy":macro_accuracy,"micro_accuracy":micro_accuracy}}
```

```python
result = {"prefix":{"micro_average":micro_average,"macro_recall":macro_recall,"macro_precision":macro_precision}}
```
